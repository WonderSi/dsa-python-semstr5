позицией stop.<br/>

Все выполненные доработки должны быть протестированы

````python
class Node:
    def __init__(self, initdata):
        self.data = initdata
        self.next = None

    def getData(self):
        return self.data

    def getNext(self):
        return self.next

    def setData(self, newdata):
        self.data = newdata

    def setNext(self, newnext):
        self.next = newnext

class UnorderedList:

    def __init__(self):
        self.head = None

    def isEmpty(self):
        return self.head == None

    def add(self, item):
        temp = Node(item)
        temp.setNext(self.head)
        self.head = temp

    def size(self):
        current = self.head
        count = 0
        while current != None:
            count = count + 1
            current = current.getNext()

        return count

    def search(self, item):
        current = self.head
        found = False
        while current != None and not found:
            if current.getData() == item:
                found = True
            else:
                current = current.getNext()

        return found

    def remove(self, item):
        current = self.head
        previous = None
        found = False
        while not found:
            if current.getData() == item:
                found = True
            else:
                previous = current
                current = current.getNext()

        if previous == None:
            self.head = current.getNext()
        else:
            previous.setNext(current.getNext())
    
    def append(self, item):
        temp = Node(item)
        if self.head is None:
            self.head = temp
        else:
            current = self.head
            while current.getNext() is not None:
                current = current.getNext()
            current.setNext(temp)

    def index(self, item):
        current = self.head
        pos = 0
        while current is not None:
            if current.getData() == item:
                return pos
            current = current.getNext()
            pos += 1
        raise ValueError("Item not in list")
    
    def insert(self, pos, item):
        if pos < 0 or pos > self.size():
            raise IndexError("Index out in range")
        if pos == 0:
            self.add(item)
        else:
            temp = Node(item)
            current = self.head
            previous = None
            current_pos = 0
            while current_pos < pos:
                previous = current
                current = current.getNext()
                current_pos += 1
            previous.setNext(temp)
            temp.setNext(current)

    def pop(self):
        if self.head is None:
            raise IndexError("pop from empty list")
        if self.head.getNext() is None:
            data = self.head.getData()
            self.head = None
            return data
        current = self.head
        previous = None
        while current.getNext() is not None:
            previous = current
            current = current.getNext()
        previous.setNext(None)
        return current.getNext()
    
    def pop(self, pos=None):
        if pos is None:
            if self.head is None:
                raise IndexError("pop from empty list")
            if self.head.getNext() is None:
                data = self.head.getData()
                self.head = None
                return data
            current = self.head
            previous = None
            while current.getNext() is not None:
                previous = current
                current = current.getNext()
            previous.setNext(None)
            return current.getData()
        else:
            if pos < 0 or pos >= self.size():
                raise IndexError("pop index out of range")
            if pos == 0:
                data = self.head.getData()
                self.head = self.head.getNext()
                return data
            current = self.head
            previous = None
            current_pos = 0
            while current_pos < pos:
                previous =current
                current = current.getNext()
                current_pos += 1
            previous.setNext(current.getNext())
            return current.getData()

    def __str__(self):
        elements = []
        current = self.head
        while current is not None:
            elements.append(repr(current.getData()))
            current = current.getNext()
        return "[" + ",".join(elements) + "]"

    def slice(self, start, stop):
        if start < 0:
            start = 0
        if stop > self.size():
            stop = self.size()
        if start >= stop or start >= self.size():
            return UnorderedList()
        
        sliced_list = UnorderedList()
        current = self.head
        pos = 0
        while pos < start:
            current = current.getNext()
            pos += 1
        while pos < stop and current is not None:
            sliced_list.append(current.getData())
            current = current.getNext()
            pos += 1
        return sliced_list

if __name__ == "__main__":
    mylist = UnorderedList()

    mylist.add(31)
    mylist.add(77)
    mylist.add(17)
    mylist.add(93)
    mylist.add(26)
    mylist.add(54)
    print(mylist)

    mylist.append(100)
    print(mylist)

    print(mylist.index(17))
    print(mylist.index(100))

    mylist.insert(2, 88)
    print(mylist)

    last = mylist.pop()
    print(last)
    print(mylist)

    removed = mylist.pop(0)
    print(removed)
    print(mylist)

    sublist = mylist.slice(1, 4)
    print(sublist)

    empty_slice = mylist.slice(20,25)
    print(empty_slice)

    full_slice = mylist.slice(0, mylist.size())
    print(full_slice)

    try:
        mylist.index(999)
    except ValueError as e:
        print(e)

    try:
        mylist.pop(100)
    except IndexError as e:
        print(e)
````

````python
def move_to_front(n):
    lst = UnorderedList()
    
    for _ in range(n):
        line = input().strip()
        if lst.search(line):
            lst.remove(line)
        lst.add(line)
    
    return lst

result = move_to_front(5)
print(result)
````

этой задачи: 1) итеративный и 2) рекурсивный.

````python
def reverse_iterative(head):
    prev = None
    current = head
    while current is not None:
        next_node = current.getNext()
        current.setNext(prev)
        prev = current
        current = next_node
    return prev

def reverse_recursive(head):
    if head is None or head.getNext() is None:
        return head

    rest = reverse_recursive(head.getNext())

    head.getNext().setNext(head)
    head.setNext(None)

    return rest

if __name__ == "__main__":
    mylist = UnorderedList()
    for x in [5, 4, 3, 2, 1]:
        mylist.add(x)

    print("До разворота:", mylist)

    mylist.head = reverse_iterative(mylist.head)

    print("После итеративного разворота:", mylist)


    mylist2 = UnorderedList()
    for x in ['x', 'y', 'z']:
        mylist2.add(x)

    print("До рекурсивного разворота:", mylist2)

    mylist2.head = reverse_recursive(mylist2.head)
    
    print("После рекурсивного разворота:", mylist2)
````

````python
class LinkedListStack:
    def __init__(self):
        self._list = UnorderedList()

    def isEmpty(self):
        return self._list.isEmpty()

    def push(self, item):
        self._list.add(item)

    def pop(self):
        if self.isEmpty():
            raise IndexError("pop from empty stack")
        top = self._list.head.getData()
        self._list.remove(top)
        return top

    def peek(self):
        if self.isEmpty():
            raise IndexError("peek from empty stack")
        return self._list.head.getData()

    def size(self):
        return self._list.size()

    def __str__(self):
        items = []
        current = self._list.head
        while current:
            items.append(repr(current.getData()))
            current = current.getNext()
        return f"Stack([{', '.join(items)}])"
    
class LinkedListQueue:
    def __init__(self):
        self._list = UnorderedList()

    def isEmpty(self):
        return self._list.isEmpty()

    def enqueue(self, item):
        self._list.append(item)

    def dequeue(self):
        if self.isEmpty():
            raise IndexError("dequeue from empty queue")
        front = self._list.head.getData()
        self._list.remove(front)
        return front

    def size(self):
        return self._list.size()

    def __str__(self):
        items = []
        current = self._list.head
        while current:
            items.append(repr(current.getData()))
            current = current.getNext()
        return f"Queue([{', '.join(items)}])"
    
class LinkedListDeque:
    def __init__(self):
        self._list = UnorderedList()

    def isEmpty(self):
        return self._list.isEmpty()

    def addFront(self, item):
        self._list.add(item)

    def addRear(self, item):
        self._list.append(item)

    def removeFront(self):
        if self.isEmpty():
            raise IndexError("removeFront from empty deque")
        front = self._list.head.getData()
        self._list.remove(front)
        return front

    def removeRear(self):
        if self.isEmpty():
            raise IndexError("removeRear from empty deque")
        current = self._list.head
        if current.getNext() is None:
            self._list.head = None
            return current.getData()
        while current.getNext().getNext() is not None:
            current = current.getNext()
        rear = current.getNext().getData()
        current.setNext(None)
        return rear

    def size(self):
        return self._list.size()

    def __str__(self):
        items = []
        current = self._list.head
        while current:
            items.append(repr(current.getData()))
            current = current.getNext()
        return f"Deque([{', '.join(items)}])"
    
if __name__ == "__main__":
    s = LinkedListStack()
    s.push(1)
    s.push(2)
    s.push(3)
    print("Стек:", s)
    print("Peek:", s.peek())
    print("Pop:", s.pop())
    print("Стек после pop:", s)

    print("\n" + "="*40 + "\n")

    q = LinkedListQueue()
    q.enqueue('a')
    q.enqueue('b')
    q.enqueue('c')
    print("Очередь:", q)
    print("Dequeue:", q.dequeue())
    print("Очередь после dequeue:", q)

    print("\n" + "="*40 + "\n")

    d = LinkedListDeque()
    d.addFront(10)
    d.addRear(20)
    d.addFront(5)
    print("Дек:", d)
    print("Remove front:", d.removeFront())
    print("Remove rear:", d.removeRear())
    print("Дек после удалений:", d)
    
````

Реализуйте класс DoubleList для построения двухсвязных списков, в которых каждый узел
содержит ссылку на узел, предшествующий ему в списке, и ссылку на следующий узел.<br/>
Класс должен иметь следующие методы: проверка списка на пустоту, поиск элемента,
размер списка, вставка в начало, вставка в конец, удаление из начала, удаление из конца,
вставка перед указанным узлом, вставка после указанного узла, удаление указанного узла.<br/>
Проведите тестирование методов реализованного класса.

````python
class DoubleNode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None

    def getData(self):
        return self.data

    def getPrev(self):
        return self.prev

    def getNext(self):
        return self.next

    def setData(self, data):
        self.data = data

    def setPrev(self, prev):
        self.prev = prev

    def setNext(self, next_node):
        self.next = next_node


class DoubleList:
    def __init__(self):
        self.head = None
        self.tail = None

    def isEmpty(self):
        return self.head is None

    def size(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.getNext()
        return count

    def search(self, item):
        current = self.head
        while current:
            if current.getData() == item:
                return True
            current = current.getNext()
        return False

    def addFront(self, item):
        new_node = DoubleNode(item)
        if self.isEmpty():
            self.head = self.tail = new_node
        else:
            new_node.setNext(self.head)
            self.head.setPrev(new_node)
            self.head = new_node

    def addRear(self, item):
        new_node = DoubleNode(item)
        if self.isEmpty():
            self.head = self.tail = new_node
        else:
            new_node.setPrev(self.tail)
            self.tail.setNext(new_node)
            self.tail = new_node

    def removeFront(self):
        if self.isEmpty():
            raise IndexError("remove_front(): список пуст")
        data = self.head.getData()
        if self.head == self.tail:
            self.head = self.tail = None
        else:
            self.head = self.head.getNext()
            self.head.setPrev(None)
        return data

    def removeRear(self):
        if self.isEmpty():
            raise IndexError("remove_rear(): список пуст")
        data = self.tail.getData()
        if self.head == self.tail:
            self.head = self.tail = None
        else:
            self.tail = self.tail.getPrev()
            self.tail.setNext(None)
        return data

    def insertBefore(self, ref_data, item):
        if self.isEmpty():
            raise ValueError("insertBefore(): список пуст")

        current = self.head
        while current and current.getData() != ref_data:
            current = current.getNext()

        if current is None:
            raise ValueError(f"insertBefore(): элемент {ref_data!r} не найден")

        new_node = DoubleNode(item)
        prev_node = current.getPrev()

        new_node.setNext(current)
        new_node.setPrev(prev_node)
        current.setPrev(new_node)

        if prev_node:
            prev_node.setNext(new_node)
        else:
            self.head = new_node

    def insertAfter(self, ref_data, item):
        if self.isEmpty():
            raise ValueError("insertAfter(): список пуст")

        current = self.head
        while current and current.getData() != ref_data:
            current = current.getNext()

        if current is None:
            raise ValueError(f"insertAfter(): элемент {ref_data!r} не найден")

        new_node = DoubleNode(item)
        next_node = current.getNext()

        new_node.setPrev(current)
        new_node.setNext(next_node)
        current.setNext(new_node)

        if next_node:
            next_node.setPrev(new_node)
        else:
            self.tail = new_node

    def remove(self, item):
        if self.isEmpty():
            raise ValueError("remove(): список пуст")

        current = self.head
        while current and current.getData() != item:
            current = current.getNext()

        if current is None:
            raise ValueError(f"remove(): элемент {item!r} не найден")

        prev_node = current.getPrev()
        next_node = current.getNext()

        if prev_node:
            prev_node.setNext(next_node)
        else:
            self.head = next_node

        if next_node:
            next_node.setPrev(prev_node)
        else:
            self.tail = prev_node

    def __str__(self):
        elements = []
        current = self.head
        while current:
            elements.append(repr(current.getData()))
            current = current.getNext()
        return "[" + ", ".join(elements) + "]"

    def __repr__(self):
        return f"DoubleList({self})"
    
if __name__ == "__main__":
    dl = DoubleList()
    print("Пустой список:", dl)
    print("Пустой?", dl.isEmpty())

    dl.addFront(10)
    dl.addRear(20)
    dl.addFront(5)
    dl.addRear(30)
    print("\nПосле addFront(10), addRear(20), addFront(5), addRear(30):")
    print("Список:", dl)
    print("Размер:", dl.size())
    print("Есть 20?", dl.search(20))
    print("Есть 100?", dl.search(100))

    dl.insertBefore(20, 15)
    print("\nПосле insertBefore(20, 15):", dl)

    dl.insertAfter(20, 25)
    print("После insertAfter(20, 25):", dl)

    front = dl.removeFront()
    rear = dl.removeRear()
    print(f"\nУдалили спереди: {front}, сзади: {rear}")
    print("Список:", dl)

    dl.remove(15)
    print("После remove(15):", dl)

    try:
        dl.remove(999)
    except ValueError as e:
        print("Ошибка:", e)

    single = DoubleList()
    single.addFront(42)
    print("\nОдиночный элемент:", single)
    print("Удалили:", single.removeFront())
    print("Теперь пустой?", single.isEmpty())
````

В HTML у каждого тега существуют открывающая и закрывающая формы, которые должны
быть сбалансированы, чтобы адекватно описывать веб-документ. Вот пример очень
простого веб-документа:
```
<html>

        </title>
    </head>
    <body>
        <h1>Hello, world</h1>
    </body>

```
Напишите программу, которая, используя структуру данных стек, проверит HTML
документ на соответствие открывающих и закрывающих тегов.

За основу решения можно взять алгоритм проверки скобочной последовательности из
лекционных материалов

````python
class Stack:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[-1]

    def size(self):
        return len(self.items)


def html_tag_checker(html_string):
    s = Stack()
    i = 0
    n = len(html_string)

    while i < n:
        if html_string[i] == '<':
            j = html_string.find('>', i)
            if j == -1:
                return False

            raw_tag = html_string[i+1:j].strip()
            if not raw_tag:
                return False

            is_self_closing = raw_tag.endswith('/')
            if is_self_closing:
                tag_content = raw_tag[:-1].strip()
                if not tag_content:
                    return False
                i = j + 1
                continue

            if raw_tag.startswith('/'):
                tag_name = raw_tag[1:].split()[0]
                if s.isEmpty():
                    return False
                if s.pop() != tag_name:
                    return False
            else:
                tag_name = raw_tag.split()[0]
                s.push(tag_name)

            i = j + 1
        else:
            i += 1

    return s.isEmpty()


html_ok = """
<html>
  <head>
    <title>Test</title>
  </head>
  <body>
    <h1>Hello</h1>
    <br/>
    <img src="pic.jpg"/>
  </body>
</html>
"""
print("Корректный HTML:", html_tag_checker(html_ok))

html_bad1 = "<div><p></div></p>"
print("Неправильная вложенность:", html_tag_checker(html_bad1))

html_bad2 = "<div></div></p>"
print("Лишний закрывающий:", html_tag_checker(html_bad2))

html_bad3 = "<div><p></div>"
print("Незакрытый тег:", html_tag_checker(html_bad3))

html_self = "<br><img src='x'><hr/>"
print("Только самозакрывающиеся:", html_tag_checker(html_self))
````

Стековый калькулятор - это устройство (реальное или виртуальное), которое вычисляет
значения арифметических выражений, записанных в постфиксной форме (также она носит
название Обратная польская запись — Википедия (wikipedia.org)). Стековым он называется
так как действует по следующим правилам: если нам встречается число - мы должны
поместить его в стек; если знак операции - мы должны достать из стека столько чисел,
сколько необходимо для выполнения данной операции и выполнить ее.

Именно так и работает т.н. математический сопроцессор (например, в INtel8086+AMD64)-
часть ЦП, отвечающая за операции над числами с плавающей запятой.

Реализуйте стековый калькулятор. Напишите программу, которая читает выражение в
обратной польской нотации и выводит получившееся значение или пишет, что выражение
составлено не корректно (если оно некорректно).

````python
def rpn_calculator(expression):
    if not expression.strip():
        return "Некорректное выражение"

    stack = []
    tokens = expression.split()
    operators = {'+', '-', '*', '/'}

    for token in tokens:
        if token in operators:
            if len(stack) < 2:
                return "Некорректное выражение"
            b = stack.pop()
            a = stack.pop()
            try:
                if token == '+':
                    res = a + b
                elif token == '-':
                    res = a - b
                elif token == '*':
                    res = a * b
                elif token == '/':
                    if b == 0:
                        return "Деление на ноль"
                    res = a / b
                stack.append(res)
            except (TypeError, ValueError):
                return "Некорректное выражение"
        else:
            try:
                num = float(token)
                stack.append(num)
            except ValueError:
                return "Некорректное выражение"

    if len(stack) != 1:
        return "Некорректное выражение"

    return stack[0]

print(rpn_calculator("3 4 + 2 * 7 /"))

print(rpn_calculator("-3.5 2 +"))
print(rpn_calculator("10 3 /"))

print(rpn_calculator("3 4 + +"))
print(rpn_calculator("3 0 /"))
print(rpn_calculator("3 4 + 5"))
print(rpn_calculator("abc 3 +"))
print(rpn_calculator(""))
````
